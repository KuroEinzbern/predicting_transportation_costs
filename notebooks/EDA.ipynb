```python
import pandas as pd
import predicting_transportation_costs.config as cfg
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
df= pd.read_csv(cfg.DATA_DIR / "train-test.csv")
print(len(df))
df.head(10)
```

#my first concern is that the data may not be IID due temporal dependency.

```python
#Seems like "delivery_lat" is not present in the print of describe().
df["delivery_lat"].dtype 

#The reason is an incorrect type. But casting directly to the same type as other geospatial features causes an error due to a corrupted value "28.6137   4"".
#so let's proceed droping the non numeric values before casting
df["delivery_lat"] = pd.to_numeric(df["delivery_lat"], errors="coerce").astype("float64")
df = df[df["delivery_lat"].notna()]
print(f"{len(df)} rows left")


```

```python
#other columns are also persisted with wrong type
print(df.dtypes)
```

```python
#correcting data types.
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["pickup"] = df["pickup"].astype("category")
df["delivery"] = df["delivery"].astype("category")
df["equipment"] = df["equipment"].astype("category")
df["load_id"] = df["load_id"].astype("string")
print(df.dtypes)

```

```python

df_daily = df.groupby("date")["posted_rate"].mean().reset_index()
df_daily = df_daily.sort_values("date")

df_daily["rolling_mean_7d"] = df_daily["posted_rate"].rolling(window=7).mean()
df_daily["rolling_mean_30d"] = df_daily["posted_rate"].rolling(window=30).mean()

fig, ax = plt.subplots(figsize=(12, 5))


ax.plot(df_daily["date"], df_daily["posted_rate"], alpha=0.3, color="gray", label="daily mean")
ax.plot(df_daily["date"], df_daily["rolling_mean_7d"], color="tab:blue", linewidth=1.5, label="Weekly mean")
ax.plot(df_daily["date"], df_daily["rolling_mean_30d"], color="tab:red", linewidth=2, label="Monthly mean")

ax.set_title("Transportation Cost Over Time (Trend & Moving Average)")
ax.set_xlabel("Date")
ax.set_ylabel("Transportation Cost")

ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
fig.autofmt_xdate()

ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
```

```python
df_daily = df.groupby("date")["posted_rate"].mean().reset_index()
df_daily = df_daily.sort_values("date")
time_series = df_daily['posted_rate']

fig, ax = plt.subplots(figsize=(12, 5))
plot_acf(time_series, lags=30, ax=ax, alpha=0.05)

ax.set_title("Autocorrelation - Daily prices")
ax.set_xlabel("Lag in days")
ax.set_ylabel("Autocorrelation coefficient")
plt.grid(True, alpha=0.3)
plt.show()
```

This confirms autocorrelation. Therefore, in order to avoid data leakage, a temporal split will be implemented, reserving 20% of the most recent data for testing and the rest for training.
